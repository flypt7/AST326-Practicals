```python
import numpy as np
import PIL
from PIL import Image
import matplotlib.pyplot as plt
import scipy.stats as stat
```

```python
wavelength = []
brightness = []

i = 0 # metadata skipper

with open("H/data/USB2G560841_18-39-30-868.txt") as f:
    for line in f: 
        if i < 13:
            i += 1
            continue # skip reading metadata lines

        (w, b) = line.strip('\n').split('\t') # remove newline, tab-separated values.
        
        wavelength.append(int(w))
        brightness.append(float(b.replace(',', '.'))) # data recorded in French locale (comma-separated decimals)
        
fig, ax = plt.subplots(1)

ax.set_ylim(0, 4200)
ax.set_xlim(350, 1050)
ax.plot(wavelength, brightness)
```
